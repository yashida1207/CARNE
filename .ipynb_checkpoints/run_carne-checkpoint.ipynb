```python
## import libraries 

import math
import numpy as np

from carne import carne
from carne import utils
```

```python
## load and tailor data 

enu = np.linspace(50, 0, 50)  # MeV 

xsec = utils.load_ibd_xsec(enu)  # cm^2
flux = utils.load_dsnb_flux(enu)  # cm^-2 s^-1 MeV^-1

sig_wc, sigeff_wc = utils.load_sig("wc")
bkg_wc = utils.load_bkg("wc")

sig_ls, sigeff_ls = utils.load_sig("ls")
bkg_ls = utils.load_bkg("ls")
```

```python
## setup carne  

hk = carne.wc(size=189., runtime=10., sig_wc, sigeff_wc, bkg_wc)
toy_hk = hk.create_toy(ntoy=10000, mode="bkgonly")

juno = carne.ls(size=14.7, runtime=10., sig_ls, sigeff_ls, bkg_ls)
toy_juno = juno.create_toy(ntoy=10000, mode="bkgonly")
```

```python
## perform fit 

scan_fsig = np.linspace(100, 0, 1)
scan_fbkg = np.linspace(100, 0, 1)
scan_set = [scan_fsig, scan_fbkg]

loglikeli_hk, teststat_hk = utils.calc_likelihood(hk, toy_hk, scan_set)
loglikeli_juno, teststat_juno = utils.calc_likelihood(juno, toy_juno, scan_set)
loglikeli_comb, teststat_comb = utils.calc_likelihood(hk, toy_hk, juno, toy_juno, scan_set)
```

```python
## run example (no touch on utils.py)

enu = np.linspace(25, 25., 50.)  # MeV 

hk = carne.wc(size=189., runtime=10., enu=enu, dsnb_model="{datadir}/dsnb_flux_ashida2023.dat")
hk.set_syst() 

toy_hk = hk.create_toy(ntoy=10000, mode="bkgonly")

scan_fsig = np.linspace(100, 0., 1.)
scan_fbkg = np.linspace(100, 0., 1.)
scan_set = [scan_fsig, scan_fbkg]
loglikeli_hk, teststat_hk = carne.calc_likelihood(hk, toy_hk, scan_set)
```
