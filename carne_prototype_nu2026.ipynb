```python
"""
    3. calculate_TS.py

        * Initial version [Feb 12, 2025, Y. Ashida]
        * Google Colab version [May 16, 2026, K. Nakazato]
        * Google Colab version2 [Jun 5, 2026, K. Nakazato]
"""

import numpy as np
from scipy.special import gammaln
from tqdm import tqdm
import matplotlib
from matplotlib.ticker import FuncFormatter, MaxNLocator
from matplotlib.ticker import LogLocator, AutoMinorLocator
from matplotlib import gridspec
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
from matplotlib.colors import LogNorm
plt.rcParams["axes.prop_cycle"] = plt.cycler("color", plt.cm.plasma(np.linspace(0.9,0.0,5)))

titlefontsize = 18
labelfontsize = 18
legendfontsize = 13
labelpad = 15

#from google.colab import drive
#drive.mount('/content/drive')

#-----------------------------------------------------------------------------------

def calc_likelihood(enu_obs, enu_pdf, pdf_sig, pdf_bkg, n_sig, n_bkg, n_bkg_nom):

    #--------------------------------------------------
    # enu_obs が 1 toy か、複数 toy かを判定
    #--------------------------------------------------

    is_single_toy = np.ndim(enu_obs) == 1

    if is_single_toy:
        enu_obs_list = [np.asarray(enu_obs)]
    else:
        enu_obs_list = enu_obs

    #--------------------------------------------------
    # n_sig, n_bkg が scalar か array かを判定
    #--------------------------------------------------

    is_single_n_sig = np.ndim(n_sig) == 0
    is_single_n_bkg = np.ndim(n_bkg) == 0

    n_sig_arr = np.atleast_1d(n_sig)
    n_bkg_arr = np.atleast_1d(n_bkg)

    # shape = (n_sig_grid, 1)
    n_sig_grid = n_sig_arr[:, None]

    # shape = (1, n_bkg_grid)
    n_bkg_grid = n_bkg_arr[None, :]

    #--------------------------------------------------
    # toy data をまとめて扱う形に変換
    #--------------------------------------------------

    n_obs_each = np.array([len(toy_data) for toy_data in enu_obs_list])
    enu_obs_flat = np.concatenate(enu_obs_list)

    n_toy = len(enu_obs_list)
    toy_index = np.repeat(np.arange(n_toy), n_obs_each)

    #--------------------------------------------------
    # likelihood の計算のメイン
    #--------------------------------------------------

    # 各観測イベントがどの energy bin に入るかを一括で探す
    bin_index = np.searchsorted(enu_pdf, enu_obs_flat, side="right") - 1

    # Poisson likelihood のうち、イベントごとに積をとる項
    # （enu_pdf の範囲外にあるイベントの確率は、一旦、ゼロとする）
    valid = (bin_index >= 0) & (bin_index < len(enu_pdf) - 1)
    prob_sig = np.zeros(len(enu_obs_flat))
    prob_bkg = np.zeros(len(enu_obs_flat))
    prob_sig[valid] = pdf_sig[bin_index[valid]]
    prob_bkg[valid] = pdf_bkg[bin_index[valid]]
    event_density = n_sig_grid[None, :, :] * prob_sig[:, None, None] + n_bkg_grid[None, :, :] * prob_bkg[:, None, None]

    # 確率密度ゼロ、またはエネルギー範囲外のイベントには非常に小さい likelihood を与える
    log_event_density = np.full(event_density.shape, -1000.0)   # temporary (need to be checked)
    good = event_density > 0
    log_event_density[good] = np.log(event_density[good])

    # 各イベントの log-likelihood 寄与を、toy_index を使って対応する toy dataset に足し込む
    #     例: toy_index = [0,0,1,1,1,2]
    #     → event 0,1 は toy 0 に加算
    #     → event 2,3,4 は toy 1 に加算
    #     → event 5 は toy 2 に加算
    likeli_each = np.zeros((n_toy, len(n_sig_arr), len(n_bkg_arr)))
    np.add.at(likeli_each, toy_index, log_event_density)

    # Poisson likelihood のうち、イベント総数に関する項
    likeli_each += -(n_sig_grid[None, :, :] + n_bkg_grid[None, :, :]) - gammaln(n_obs_each[:, None, None] + 1)   # gammaln(N+1) = log(N!)

    likeli_each += -0.5 * ((n_bkg_grid[None, :, :] - n_bkg_nom) / (n_bkg_nom * 0.2))**2   # pull term with a temporary 20% syst

    #--------------------------------------------------
    # 出力の形を整える
    #--------------------------------------------------

    if is_single_toy:
        likeli_each = likeli_each[0]
    else:
        likeli_each = np.moveaxis(likeli_each, 0, -1)
        # shape = (n_sig_grid, n_bkg_grid, n_toy)

    if is_single_n_sig and is_single_n_bkg:
        return np.squeeze(likeli_each)

    elif is_single_n_sig:
        return np.squeeze(likeli_each, axis=0)

    elif is_single_n_bkg:
        return np.squeeze(likeli_each, axis=1)

    else:
        return likeli_each   # log-likelihood

#-----------------------------------------------------------------------------------
## calculate likelihood and TS
use_wc = configs["use_wc"]
use_ls = configs["use_ls"]

#if use_wc:
#    pdf_arr_wc = np.load(f"/content/drive/MyDrive/wc_pdf.npz")
#    enu_wc = pdf_arr_wc["enu_wc"]
#    pdf_sig_wc = pdf_arr_wc["pdf_sig_wc"]
#    pdf_bkg_tot_wc = pdf_arr_wc["pdf_bkg_tot_wc"]
#    n_sig_wc = int(pdf_arr_wc["n_evt_wc"][0])
#    n_bkg_tot_wc = int(pdf_arr_wc["n_evt_wc"][1])
#    toy_arr_wc = np.load(f"/content/drive/MyDrive/toy_data_wc_{toyID}.npz")
#    toy_data_wc = toy_arr_wc["toy_data_wc"]
#
#if use_ls:
#    pdf_arr_ls = np.load(f"/content/drive/MyDrive/ls_pdf.npz")
#    enu_ls = pdf_arr_ls["enu_ls"]
#    pdf_sig_ls = pdf_arr_ls["pdf_sig_ls"]
#    pdf_bkg_tot_ls = pdf_arr_ls["pdf_bkg_tot_ls"]
#    n_sig_ls = int(pdf_arr_ls["n_evt_ls"][0])
#    n_bkg_tot_ls = int(pdf_arr_ls["n_evt_ls"][1])
#    toy_arr_ls = np.load(f"/content/drive/MyDrive/toy_data_ls_{toyID}.npz")
#    toy_data_ls = toy_arr_ls["toy_data_ls"]

n_sig_scale_list = np.linspace(0.00001, 1.20, 121)
n_bkg_wc_scale_list = np.linspace(0.80, 1.20, 81)
n_bkg_ls_scale_list = np.linspace(0.50, 1.30, 81)

"""
likeli_arr = []  # log-likelihood

for i, n_sig_scale in enumerate(n_sig_scale_list):
    likeli_arr.append([])

    for j, n_bkg_wc_scale in enumerate(n_bkg_wc_scale_list):
        likeli_arr[i].append([])

        for k, n_bkg_ls_scale in enumerate(n_bkg_ls_scale_list):

            ## WC part
            if use_wc:
                likeli_wc_tmp = calc_likelihood(enu_obs=toy_data_wc, enu_pdf=enu_wc,
                                                pdf_sig=pdf_sig_wc, pdf_bkg=pdf_bkg_tot_wc,
                                                n_sig=n_sig_wc*n_sig_scale,
                                                n_bkg=n_bkg_tot_wc*n_bkg_wc_scale,
                                                n_bkg_nom=int(n_bkg_tot_wc))
            else:
                likeli_wc_tmp = 0.0

            ## LS part
            if use_ls:
                likeli_ls_tmp = calc_likelihood(enu_obs=toy_data_ls, enu_pdf=enu_ls,
                                                pdf_sig=pdf_sig_ls, pdf_bkg=pdf_bkg_tot_ls,
                                                n_sig=n_sig_ls*n_sig_scale,
                                                n_bkg=n_bkg_tot_ls*n_bkg_ls_scale,
                                                n_bkg_nom=int(n_bkg_tot_ls))
            else:
                likeli_ls_tmp = 0.0

            ## combined
            #print(likeli_wc_tmp, likeli_ls_tmp)
            likeli_tmp = likeli_wc_tmp + likeli_ls_tmp
            likeli_arr[i][j].append(likeli_tmp)

likeli_arr = np.array(likeli_arr)
likeli_max = np.amax(likeli_arr)
ts_arr = -2.*(likeli_arr - likeli_max)
"""
#-----------------------------------------------
# toy data を複数セット作った場合
#-----------------------------------------------
f_sig_limit_wc_arr = []
f_sig_limit_ls_arr = []
f_sig_limit_comb_arr = []

if use_wc:
    n_toy = len(toy_data_wc_all)
else:
    n_toy = len(toy_data_ls_all)

for id in tqdm(range(n_toy)):
    toyID = str(id).zfill(6)

    ## WC part
    if use_wc:
        likeli_wc_arr = calc_likelihood(enu_obs=toy_data_wc_all[id], enu_pdf=enu_wc,
                                        pdf_sig=pdf_sig_wc, pdf_bkg=pdf_bkg_tot_wc,
                                        n_sig=n_sig_wc * n_sig_scale_list,
                                        n_bkg=n_bkg_tot_wc * n_bkg_wc_scale_list,
                                        n_bkg_nom=int(n_bkg_tot_wc))
    else:
        likeli_wc_arr = np.zeros(len(n_sig_scale_list), len(n_bkg_wc_scale_list))

    ## LS part
    if use_ls:
        likeli_ls_arr = calc_likelihood(enu_obs=toy_data_ls_all[id], enu_pdf=enu_ls,
                                        pdf_sig=pdf_sig_ls, pdf_bkg=pdf_bkg_tot_ls,
                                        n_sig=n_sig_ls * n_sig_scale_list,
                                        n_bkg=n_bkg_tot_ls * n_bkg_ls_scale_list,
                                        n_bkg_nom=int(n_bkg_tot_ls))

    else:
        likeli_ls_arr = np.zeros(len(n_sig_scale_list), len(n_bkg_ls_scale_list))

    ## combined
    likeli_arr = likeli_wc_arr[:, :, None] + likeli_ls_arr[:, None, :]
    likeli_max = np.amax(likeli_arr)
    ts_arr = -2.0 * (likeli_arr - likeli_max)

    # 各 signal scale i について、
    # background nuisance parameters j, k を動かし、
    # TS が最小になる点を探す
    ts_comb_arr = np.min(ts_arr, axis=(1, 2))

    # profile likelihood に基づく best-fit signal scale
    i_comb_min = np.argmin(ts_comb_arr)
    f_sig_bestfit_comb = n_sig_scale_list[i_comb_min]

    ts_1sigma = 1.00; ts_2sigma = 3.84; ts_3sigma = 6.63

    f_sig_1sigma_comb = np.max(n_sig_scale_list[ts_comb_arr < ts_1sigma])
    f_sig_2sigma_comb = np.max(n_sig_scale_list[ts_comb_arr < ts_2sigma])
    f_sig_3sigma_comb = np.max(n_sig_scale_list[ts_comb_arr < ts_3sigma])
    f_sig_limit_arr = np.array([f_sig_bestfit_comb, f_sig_1sigma_comb, f_sig_2sigma_comb, f_sig_3sigma_comb])

    f_sig_limit_comb_arr.append(f_sig_limit_arr)

    if use_wc:
        # 各 signal scale i, WC background scale j について、
        # LS background nuisance parameter k を動かし、
        # TS が最小になる点を探す
        ts_wc_arr = np.min(ts_arr, axis=2).T

        F_SIG_WC, N_BKG_WC = np.meshgrid(n_sig_scale_list, n_bkg_tot_wc*n_bkg_wc_scale_list)

        # TS regions
        mask_bestfit = (ts_wc_arr == np.min(ts_wc_arr))
        mask_1sigma = (ts_wc_arr < ts_1sigma)
        mask_2sigma = ((ts_wc_arr >= ts_1sigma) & (ts_wc_arr < ts_2sigma))
        mask_3sigma = ((ts_wc_arr >= ts_2sigma) & (ts_wc_arr < ts_3sigma))

        f_sig_1sigma_wc_arr = F_SIG_WC[mask_1sigma]; f_bkg_1sigma_wc_arr = N_BKG_WC[mask_1sigma]/n_bkg_tot_wc
        f_sig_2sigma_wc_arr = F_SIG_WC[mask_2sigma]; f_bkg_2sigma_wc_arr = N_BKG_WC[mask_2sigma]/n_bkg_tot_wc
        f_sig_3sigma_wc_arr = F_SIG_WC[mask_3sigma]; f_bkg_3sigma_wc_arr = N_BKG_WC[mask_3sigma]/n_bkg_tot_wc
        f_sig_bestfit_wc = F_SIG_WC[mask_bestfit]; f_bkg_bestfit_wc = N_BKG_WC[mask_bestfit]/n_bkg_tot_wc

        f_sig_1sigma_wc = np.max(f_sig_1sigma_wc_arr)
        f_sig_2sigma_wc = np.max(f_sig_2sigma_wc_arr)
        f_sig_3sigma_wc = np.max(f_sig_3sigma_wc_arr)
        f_sig_limit_arr = np.array([f_sig_bestfit_wc[0], f_sig_1sigma_wc, f_sig_2sigma_wc, f_sig_3sigma_wc])

        # save
        #np.savez(f"/content/drive/MyDrive/TS_toy_data_wc_{toyID}.npz", f_sig_limit_arr=f_sig_limit_arr)
        f_sig_limit_wc_arr.append(f_sig_limit_arr)

    if use_ls:
        # 各 signal scale i, WC background scale k について、
        # WC background nuisance parameter j を動かし、
        # TS が最小になる点を探す
        ts_ls_arr = np.min(ts_arr, axis=1).T

        F_SIG_LS, N_BKG_LS = np.meshgrid(n_sig_scale_list, n_bkg_tot_ls*n_bkg_ls_scale_list)

        # TS regions
        mask_bestfit = (ts_ls_arr == np.min(ts_ls_arr))
        mask_1sigma = (ts_ls_arr < ts_1sigma)
        mask_2sigma = ((ts_ls_arr >= ts_1sigma) & (ts_ls_arr < ts_2sigma))
        mask_3sigma = ((ts_ls_arr >= ts_2sigma) & (ts_ls_arr < ts_3sigma))

        f_sig_1sigma_ls_arr = F_SIG_LS[mask_1sigma]; f_bkg_1sigma_ls_arr = N_BKG_LS[mask_1sigma]/n_bkg_tot_ls
        f_sig_2sigma_ls_arr = F_SIG_LS[mask_2sigma]; f_bkg_2sigma_ls_arr = N_BKG_LS[mask_2sigma]/n_bkg_tot_ls
        f_sig_3sigma_ls_arr = F_SIG_LS[mask_3sigma]; f_bkg_3sigma_ls_arr = N_BKG_LS[mask_3sigma]/n_bkg_tot_ls
        f_sig_bestfit_ls = F_SIG_LS[mask_bestfit]; f_bkg_bestfit_ls = N_BKG_LS[mask_bestfit]/n_bkg_tot_ls

        f_sig_1sigma_ls = np.max(f_sig_1sigma_ls_arr)
        f_sig_2sigma_ls = np.max(f_sig_2sigma_ls_arr)
        f_sig_3sigma_ls = np.max(f_sig_3sigma_ls_arr)
        f_sig_limit_arr = np.array([f_sig_bestfit_ls[0], f_sig_1sigma_ls, f_sig_2sigma_ls, f_sig_3sigma_ls])

        # save
        #np.savez(f"/content/drive/MyDrive/TS_toy_data_ls_{toyID}.npz", f_sig_limit_arr=f_sig_limit_arr)
        f_sig_limit_ls_arr.append(f_sig_limit_arr)

    """
    ## store into 2D arrays
    index_min = np.unravel_index(np.argmin(ts_arr), ts_arr.shape)
    i_min = index_min[0]
    j_min = index_min[1]
    k_min = index_min[2]

    if use_wc:
        ts_wc_arr = []
        for j, n_bkg_wc_scale in enumerate(n_bkg_wc_scale_list):
            ts_wc_arr.append([])
            for i, n_sig_scale in enumerate(n_sig_scale_list):
                ts_wc_arr[j].append(ts_arr[i][j][k_min])
        ts_wc_arr = np.array(ts_wc_arr)
        F_SIG_WC, N_BKG_WC = np.meshgrid(n_sig_scale_list, n_bkg_tot_wc*n_bkg_wc_scale_list)

        ts_1sigma = 1.00; ts_2sigma = 3.84; ts_3sigma = 6.63

        f_sig_bestfit = []; f_bkg_bestfit = []
        f_sig_1sigma_arr = []; f_bkg_1sigma_arr = []
        f_sig_2sigma_arr = []; f_bkg_2sigma_arr = []
        f_sig_3sigma_arr = []; f_bkg_3sigma_arr = []

        for j, n_bkg_wc_scale in enumerate(n_bkg_wc_scale_list):
            for i, n_sig_scale in enumerate(n_sig_scale_list):
                if ts_wc_arr[j][i] < ts_1sigma:
                    f_sig_1sigma_arr.append(n_sig_scale)
                    f_bkg_1sigma_arr.append(n_bkg_wc_scale)
                elif ts_wc_arr[j][i] < ts_2sigma:
                    f_sig_2sigma_arr.append(n_sig_scale)
                    f_bkg_2sigma_arr.append(n_bkg_wc_scale)
                elif ts_wc_arr[j][i] < ts_3sigma:
                    f_sig_3sigma_arr.append(n_sig_scale)
                    f_bkg_3sigma_arr.append(n_bkg_wc_scale)
                if ts_wc_arr[j][i] == 0:
                    f_sig_bestfit.append(n_sig_scale)
                    f_bkg_bestfit.append(n_bkg_wc_scale)

        f_sig_1sigma_wc_arr = np.array(f_sig_1sigma_arr); f_bkg_1sigma_wc_arr = np.array(f_bkg_1sigma_arr)
        f_sig_2sigma_wc_arr = np.array(f_sig_2sigma_arr); f_bkg_2sigma_wc_arr = np.array(f_bkg_2sigma_arr)
        f_sig_3sigma_wc_arr = np.array(f_sig_3sigma_arr); f_bkg_3sigma_wc_arr = np.array(f_bkg_3sigma_arr)
        f_sig_bestfit_wc = np.array(f_sig_bestfit); f_bkg_bestfit_wc = np.array(f_bkg_bestfit)

        f_sig_1sigma_wc = np.max(f_sig_1sigma_wc_arr)
        f_sig_2sigma_wc = np.max(f_sig_2sigma_wc_arr)
        f_sig_3sigma_wc = np.max(f_sig_3sigma_wc_arr)
        f_sig_limit_arr = np.array([f_sig_bestfit_wc[0], f_sig_1sigma_wc, f_sig_2sigma_wc, f_sig_3sigma_wc])

        # save
        #np.savez(f"/content/drive/MyDrive/TS_toy_data_wc_{toyID}.npz", f_sig_limit_arr=f_sig_limit_arr)
        f_sig_limit_wc_arr.append(f_sig_limit_arr)

    if use_ls:
        ts_ls_arr = []
        for k, n_bkg_ls_scale in enumerate(n_bkg_ls_scale_list):
            ts_ls_arr.append([])
            for i, n_sig_scale in enumerate(n_sig_scale_list):
                ts_ls_arr[k].append(ts_arr[i][j_min][k])
        ts_ls_arr = np.array(ts_ls_arr)
        F_SIG_LS, N_BKG_LS = np.meshgrid(n_sig_scale_list, n_bkg_tot_ls*n_bkg_ls_scale_list)

        ts_1sigma = 1.00; ts_2sigma = 3.84; ts_3sigma = 6.63

        f_sig_bestfit = []; f_bkg_bestfit = []
        f_sig_1sigma_arr = []; f_bkg_1sigma_arr = []
        f_sig_2sigma_arr = []; f_bkg_2sigma_arr = []
        f_sig_3sigma_arr = []; f_bkg_3sigma_arr = []

        for j, n_bkg_ls_scale in enumerate(n_bkg_ls_scale_list):
            for i, n_sig_scale in enumerate(n_sig_scale_list):
                if ts_ls_arr[j][i] < ts_1sigma:
                    f_sig_1sigma_arr.append(n_sig_scale)
                    f_bkg_1sigma_arr.append(n_bkg_ls_scale)
                elif ts_ls_arr[j][i] < ts_2sigma:
                    f_sig_2sigma_arr.append(n_sig_scale)
                    f_bkg_2sigma_arr.append(n_bkg_ls_scale)
                elif ts_ls_arr[j][i] < ts_3sigma:
                    f_sig_3sigma_arr.append(n_sig_scale)
                    f_bkg_3sigma_arr.append(n_bkg_ls_scale)
                if ts_ls_arr[j][i] == 0:
                    f_sig_bestfit.append(n_sig_scale)
                    f_bkg_bestfit.append(n_bkg_ls_scale)

        f_sig_1sigma_ls_arr = np.array(f_sig_1sigma_arr); f_bkg_1sigma_ls_arr = np.array(f_bkg_1sigma_arr)
        f_sig_2sigma_ls_arr = np.array(f_sig_2sigma_arr); f_bkg_2sigma_ls_arr = np.array(f_bkg_2sigma_arr)
        f_sig_3sigma_ls_arr = np.array(f_sig_3sigma_arr); f_bkg_3sigma_ls_arr = np.array(f_bkg_3sigma_arr)
        f_sig_bestfit_ls = np.array(f_sig_bestfit); f_bkg_bestfit_ls = np.array(f_bkg_bestfit)

        f_sig_1sigma_ls = np.max(f_sig_1sigma_ls_arr)
        f_sig_2sigma_ls = np.max(f_sig_2sigma_ls_arr)
        f_sig_3sigma_ls = np.max(f_sig_3sigma_ls_arr)
        f_sig_limit_arr = np.array([f_sig_bestfit_ls[0], f_sig_1sigma_ls, f_sig_2sigma_ls, f_sig_3sigma_ls])

        # save
        #np.savez(f"/content/drive/MyDrive/TS_toy_data_ls_{toyID}.npz", f_sig_limit_arr=f_sig_limit_arr)
        f_sig_limit_ls_arr.append(f_sig_limit_arr)
        """

f_sig_limit_wc_arr = np.array(f_sig_limit_wc_arr)
f_sig_limit_ls_arr = np.array(f_sig_limit_ls_arr)
f_sig_limit_comb_arr = np.array(f_sig_limit_comb_arr)

#-----------------------------------------------
# for plotting
#-----------------------------------------------

if True:  # for plotting

    #max_bestfit = np.max(f_sig_comb_bestfit)
    #toy_ids_max = np.where(f_sig_comb_bestfit == max_bestfit)[0]
    #idmax = toy_ids_max[-1]

    fig, ax = plt.subplots(1, 1, figsize=(8,8))
    ax.plot(n_sig_scale_list, ts_comb_arr, color="blue", linewidth=2.5, label="TS curve")
    ax.axvline(f_sig_bestfit_comb, color="red", linestyle="dotted", linewidth=2, label="best fit")
    ax.axhline(ts_1sigma, linestyle="dashdot", linewidth=2, label=r"$1\sigma$")
    ax.axhline(ts_2sigma, linestyle="dashdot", linewidth=2, label=r"$95\%$ CL")
    ax.axhline(ts_3sigma, linestyle="dashdot", linewidth=2, label=r"$99\%$ CL")

    ax.set_xlabel(r"$f_{sig}$", fontsize=labelfontsize, labelpad=labelpad)
    ax.set_ylabel("$TS_{comb}$", fontsize=labelfontsize, labelpad=labelpad)

    ax.set_xlim(0., np.max(n_sig_scale_list))
    ax.set_ylim(bottom=0)

    plt.legend(bbox_to_anchor=(1,1), borderaxespad=1, fontsize=legendfontsize)
    plt.tick_params(labelsize=labelfontsize)
    plt.subplots_adjust(wspace=0, hspace=0.05)
    #ax.legend()

    #plt.show()

    if use_wc:
        # 2D map
        fig, ax = plt.subplots(1, 1, figsize=(10,8))

        ax_cmap = ax.pcolormesh(F_SIG_WC, N_BKG_WC, ts_wc_arr, cmap="viridis", norm=LogNorm(vmin=0.1, vmax=np.max(ts_wc_arr)))
        cbar = fig.colorbar(ax_cmap)

        cbar.set_label(r"Test statistic", fontsize=labelfontsize, labelpad=labelpad)
        cbar.ax.tick_params(labelsize=labelfontsize)

        ax.scatter(np.array([-1]), np.array([-1]),  # dummy for legend
                   marker="s", s=50, c="white", alpha=1.0, edgecolors="black", label="Best-fit")

        ax.set_xlim(0.,n_sig_scale_list[-1])
        ax.set_ylim(n_bkg_tot_wc*n_bkg_wc_scale_list[0],n_bkg_tot_wc*n_bkg_wc_scale_list[-1])
        ax.set_xlabel(r"$f_{sig}$", fontsize=labelfontsize, labelpad=labelpad)
        ax.set_ylabel(r"$N_{bkg,wc}$", fontsize=labelfontsize, labelpad=labelpad)

        plt.legend(bbox_to_anchor=(1,1), loc="upper right", borderaxespad=1, fontsize=legendfontsize)
        plt.tick_params(labelsize=labelfontsize)
        plt.subplots_adjust(wspace=0, hspace=0.05)

        #plt.savefig(f"/content/drive/MyDrive/TS_toy_data_wc_2D.png", format="png", dpi=300, bbox_inches="tight")

        # contour map
        fig, ax = plt.subplots(1, 1, figsize=(8,8))

        ax.scatter(np.array([-1]), np.array([-1]),  # dummy for legend
                   marker="*", s=100, c="red", alpha=1.0, edgecolors="red", label="Best-fit")
        ax.scatter(f_sig_1sigma_wc_arr, f_bkg_1sigma_wc_arr,
                   marker="s", s=200, c="yellow", alpha=1.0, edgecolors="yellow", label="1\u03C3")
        ax.scatter(f_sig_2sigma_wc_arr, f_bkg_2sigma_wc_arr,
                   marker="s", s=200, c="greenyellow", alpha=1.0, edgecolors="greenyellow", label=r"$95\%$ CL")
        ax.scatter(f_sig_3sigma_wc_arr, f_bkg_3sigma_wc_arr,
                   marker="s", s=200, c="green", alpha=1.0, edgecolors="green", label=r"$99\%$ CL")
        ax.scatter(f_sig_bestfit_wc, f_bkg_bestfit_wc,
                   marker="*", s=100, c="red", alpha=1.0, edgecolors="red", label="")

        ax.set_xlim(0.,n_sig_scale_list[-1])
        ax.set_ylim(n_bkg_wc_scale_list[0],n_bkg_wc_scale_list[-1])
        ax.set_xlabel(r"$f_{sig}$", fontsize=labelfontsize, labelpad=labelpad)
        ax.set_ylabel(r"$f_{bkg,wc}$", fontsize=labelfontsize, labelpad=labelpad)

        plt.legend(bbox_to_anchor=(1,1), loc="upper right", borderaxespad=1, fontsize=legendfontsize)
        plt.tick_params(labelsize=labelfontsize)
        plt.subplots_adjust(wspace=0, hspace=0.05)

        #plt.savefig(f"/content/drive/MyDrive/contour_toy_data_wc.png", format="png", dpi=300, bbox_inches="tight")

    if use_ls:
        # 2D map
        fig, ax = plt.subplots(1, 1, figsize=(10,8))

        ax_cmap = ax.pcolormesh(F_SIG_LS, N_BKG_LS, ts_ls_arr, cmap="viridis", norm=LogNorm(vmin=0.1, vmax=np.max(ts_ls_arr)))
        cbar = fig.colorbar(ax_cmap)

        cbar.set_label(r"Test statistic", fontsize=labelfontsize, labelpad=labelpad)
        cbar.ax.tick_params(labelsize=labelfontsize)

        ax.scatter(np.array([-1]), np.array([-1]),  # dummy for legend
                   marker="s", s=50, c="white", alpha=1.0, edgecolors="black", label="Best-fit")

        ax.set_xlim(0.,n_sig_scale_list[-1])
        ax.set_ylim(n_bkg_tot_ls*n_bkg_ls_scale_list[0],n_bkg_tot_ls*n_bkg_ls_scale_list[-1])
        ax.set_xlabel(r"$f_{sig}$", fontsize=labelfontsize, labelpad=labelpad)
        ax.set_ylabel(r"$N_{bkg,ls}$", fontsize=labelfontsize, labelpad=labelpad)

        plt.legend(bbox_to_anchor=(1,1), loc="upper right", borderaxespad=1, fontsize=legendfontsize)
        plt.tick_params(labelsize=labelfontsize)
        plt.subplots_adjust(wspace=0, hspace=0.05)

        #plt.savefig(f"/content/drive/MyDrive/TS_toy_data_ls_2D.png", format="png", dpi=300, bbox_inches="tight")

        # contour map
        fig, ax = plt.subplots(1, 1, figsize=(8,8))

        ax.scatter(np.array([-1]), np.array([-1]),  # dummy for legend
                   marker="*", s=100, c="red", alpha=1.0, edgecolors="red", label="Best-fit")
        ax.scatter(f_sig_1sigma_ls_arr, f_bkg_1sigma_ls_arr,
                   marker="s", s=200, c="yellow", alpha=1.0, edgecolors="yellow", label="1\u03C3")
        ax.scatter(f_sig_2sigma_ls_arr, f_bkg_2sigma_ls_arr,
                   marker="s", s=200, c="greenyellow", alpha=1.0, edgecolors="greenyellow", label=r"$95\%$ CL")
        ax.scatter(f_sig_3sigma_ls_arr, f_bkg_3sigma_ls_arr,
                   marker="s", s=200, c="green", alpha=1.0, edgecolors="green", label=r"$99\%$ CL")
        ax.scatter(f_sig_bestfit_ls, f_bkg_bestfit_ls,
                   marker="*", s=100, c="red", alpha=1.0, edgecolors="red", label="")

        ax.set_xlim(0.,n_sig_scale_list[-1])
        ax.set_ylim(n_bkg_ls_scale_list[0],n_bkg_ls_scale_list[-1])
        ax.set_xlabel(r"$f_{sig}$", fontsize=labelfontsize, labelpad=labelpad)
        ax.set_ylabel(r"$f_{bkg,ls}$", fontsize=labelfontsize, labelpad=labelpad)

        plt.legend(bbox_to_anchor=(1,1), loc="upper right", borderaxespad=1, fontsize=legendfontsize)
        plt.tick_params(labelsize=labelfontsize)
        plt.subplots_adjust(wspace=0, hspace=0.05)

        #plt.savefig(f"/content/drive/MyDrive/contour_toy_data_ls.png", format="png", dpi=300, bbox_inches="tight")

```
